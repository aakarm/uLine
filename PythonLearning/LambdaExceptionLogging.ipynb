# Lambda Function
Small 1 line anonymous function
    
    lambda args: expression

typically used when func is used only once in the code
or 
for larger function and make the usage of that function easier


```python
#example lamdba function
add10 = lambda x: x+10

#same as the full func below
#  def add10_func(y):
#     return (y+10)

add10(4)
```

```python
mult = lambda x,y: x*y

mult(2,5)
```

```python
#example of lambda within another function

points2D = [(1,2), (15,1), (5,-3), (10,4)]
points2D_sorted = sorted(points2D)

points2D_sorted
```

```python
points2D_sorted_withArg = sorted(points2D, key=lambda x: x[1])
points2D_sorted_withArg
```

```python
points2D_sorted_bySum = sorted(points2D, key=lambda x: x[0] + x[1])
points2D_sorted_bySum
```

```python
#lamda with a map function 
#transforms each element with a function
#map(func, seq)

seq = [1,2,3,4,5,6]
# create a new list with each element multiplied by 2
newSeq = map(lambda x: x*2,seq)
list(newSeq)

#this is same as:
#c = [x*2 for x in seq]
```

```python
#Filter functions with lambda
#returns true or false for each element
filteredSeq = filter(lambda x: x%2==0, seq)
list(filteredSeq)

#same as the following:
#c = [x for x in seq if x%2==0]
```

```python
#reduce function
# takes each element and returns a single value
from functools import reduce
anotherSeq = [1,2,3]
productSeq = reduce(lambda x,y: x*y, anotherSeq)
productSeq
```

## Exceptions and Syntax errors 
Raise and handle exceptions, most common ones

```python
#syntax error
# a =5 print(a)

#type error
# a  = 5+ten
#a = 5 + '10'

#name error
# a = 5
# b =c+3

#Value error
# a = [1,2,3]
# a.remove(5)

#index error 
# a[5]

#key error for dictionary
# myDict = {'name': 'Max'}
# myDict['age']

```

```python
#Raise an exception 
x = -5
if x < 0:
    raise Exception('x should be +ve')
```

```python
#assert statement throws and assertion is not true 
assert(x>=0), 'x is not +ve'
```

```python
#try-except block
try:
    a = 5 / 0
except:
    print('error happened')
```

```python
#try-excpet with saving the excpetion and else + finally clauses
try:
    a = 5 / 1
    b = a + '10'
except ZeroDivisionError as e:
    print(e)
except TypeError as e:
    print(e)
else:
    print('everyhting is fine')
finally:
    #used for cleanup operation
    print('cleaning up...') 
```

```python
# define our own excpetion
class ValueTooHighError(Exception):
    pass

class ValueTooSmallError(Exception):
    def __init__(self, message, value):
        self.message = message
        self.value = value


def testValue(x):
    if x > 100:
        raise ValueTooHighError('Value is too high')
    if x < 5:
        raise ValueTooSmallError('Value is too small', x)
    
try:
    testValue(1)
except ValueTooHighError as e:
    print(e)
except ValueTooSmallError as e:
    print(e.message, e.value)
```
