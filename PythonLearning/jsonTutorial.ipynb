# JSON Tutorial

```python
# encoding a python dict into a JSON
import json

person = {"name": "John", "age": 30, "city": "New York", "hasChildren": False, "titles": ["engineer", "programmer"]}
personJSON = json.dumps(person, indent=4)
personJSON
```

```python
with open('person.json', 'w') as file:
    json.dump(person, file, indent=4)
```

```python
#deserialization of JSON data (string)

examplePerson = json.loads(personJSON)
examplePerson
```

```python
#load from file 
with open('person.json', 'r') as fileNew:
    personNew = json.load(fileNew)
    print(personNew)
```

```python
#Custom object / Class

class User:
    def __init__ (self, name, age):
        self.name = name
        self.age = age

user = User('AK', 27) 

#custom encoding function for json
def encodeUser(obj):
    if isinstance(obj, User):
        return {'name': obj.name, 'age': obj.age, obj.__class__.__name__: True}
    else:
        raise TypeError('Object of type User is not serializable')

userJSON = json.dumps(user, default =encodeUser)
#same as:
#userJSON = UserEncoder().encode(user)
userJSON
```

```python
#decode object back to dictionary

revertedUser = json.loads(userJSON)
revertedUser
```

```python
def decodeUser(dct):
    if User.__name__ in dct:
        return User(name=dct['name'], age=dct['age'])
    return dct

newUser = json.loads(userJSON, object_hook=decodeUser)
type(newUser)
print(newUser.name)
```
