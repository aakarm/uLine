# Function Arguments Sections
- Arguments vs parameters
- positional and keyword arguments
- default arguments
- variable length arguments (*args and **kwargs)
- container unpacking into function arguments
- local vs global arguments
- parameter passing (by value or by reference)


```python
#Arguements and paramters
#Arguments are values passed into the function for these parameters

#here 'name' is the parameter for the function
def printName(name):
    print(name) 

# here Aakar is the argument
printName("Aakar")
```

```python
# Positional or keyword arguments

def foo(a, b, c, d=4):
    print(a, b, c, d)

#positional arguments
foo(1,2,3)

#keyword arguments (order does not matter) 
foo(a=1, c=3, b=2)
foo(1, b =3, c=4)

# Default arguments- usually assigned in parameter as default
# has to be at the end 
foo(1,3, 4,7)
```

```python
# Variable length arguments
def faa(a, b, *args, **kwargs):
    #one asterix -> any arguments can be passed 
    #two asterix -> any number of key word arguments 
    print(a,b)
    #args is a tuple
    for arg in args:
        print(arg)

    #kwargs is a dictionary 
    for key in kwargs: 
        print(key, kwargs[key])

faa(1,2,3,4, six=6, seven=10)
#no args or kwargs
faa(1,2)
```

```python
# force key word arguements only 
def fim(a,b,*,c,d):
    print(a,b,c,d)

fim(1,2,c=3,d=-5)
```

```python
#unpacking a list or dictionary into a function argument
def boo(a,b,c):
    print(a,b,c)

myList = [0,1,2]
boo(*myList)

myDict = {'a':1, 'b':2, 'c':5}
boo(**myDict)
```

```python
#local vs global variables
#this is global variable
number = 0

def baa():
    x = number
    #cannot change the number variable inside, unless we use the keyword 'global'
    print('number inside func', x)

baa()
```

# parameter passing
call value and call by reference ->IN python call by object 

rules:
- reference to an object (but the reference is passed by values)
- difference between mutable and immutable data types


```python
def passFunc(x):
    x = 5

var = 10
passFunc(var)
#still print 10 as the var is an integer (immutable) and x is a local variable
print(var)
```

```python
#mutable object can be changes
def passList(aList):
    aList.append(5)
    #if I were to create the aList with all new variable, then that will rebind the variable as local
    #aList = [100,200,300] -> will now be a local vairable and not change the global list

    #we can add items like this too
    aList += [100,200]
    #but if we were to do the following, it will be taken as a local variable
    #aList = aList + [100,200]

myList = [0,1,2]
passList(myList)
#The list is mutable so it can be updated by the function 
print(myList)
```

```python
#function using multiple/variable length arguments and key word arguments 
def moo(a,b,*args, **kwargs):
    print(a, b)
    for arg in args: 
        print(arg)
    for key in kwargs:
        print(key, kwargs[key])

moo(1,3, 5,7, d= 9, m =10)
```

# Shallow and Deep Copying

- shollow copy: one level deep, only reference of nested child objects 
- deep copy: full independent copy 

```python
org = 5
cpy = org # creates a new variable with the same reference

#For mutable types like list, copy of a iterable is going to be the same as the origianl
#so the original list will change 

import copy
lst = [0,1,2,3,4]
newCpy = copy.copy(lst)
newCpy[0] = 10
# does nort affect the origianl list
print(lst)
print(newCpy)
```

```python
#nested list
firstLst = [[0,1,2,3,4],[10,12,14]]
deepCpy = copy.deepcopy(firstLst)
deepCpy[0][2] = -10

print(deepCpy)
print(firstLst)

lstCpy = copy.copy(firstLst)

lstCpy[0][2] = -10

print(lstCpy)
print(firstLst)
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p1 = Person('alex', 22)
p2 = p1

#both get affected
p2.age = 29

print(p2.age)
print(p1.age)
```

```python
p3 = copy.copy(p1)
p3.age = 25
print(p3.age)
print(p1.age)
```

# Context Managers

```python
with open('notes.txt', 'w') as file:
    file.write('to dos for ...')

# this is a typical example of context manager. the 'with' statement automativcally closes the file properly
# also same as lock()
```

```python
# enter and exit
class ManagedFile:
    def __init__(self, filename):
        print('init')
        self.filename = filename

    def __enter__(self):
        # allocate our resoruce
        print('enter')
        self.file = open(self.filename, 'w')
        return self.file
    
    def __exit__(self, exc_type, exc_value, exc_traceback):
        if self.file:
            self.file.close()
            print(exc_type)
            print('exiting')

with ManagedFile('notes.txt') as file:
    print('do some stuff')
    file.write('to dos...')
```
