# Collections in Python
Counter, namedtuple, orderedDict, defaultDict, deque

```python
#Special container
from collections import Counter, namedtuple, OrderedDict, defaultdict, deque
```

```python
a = 'aaaabbbccccc'
#count the number of elements and then put in a dictionary
myCounter = Counter(a)
myCounter.values()
```

```python
countDict = myCounter.most_common(2)
countDict[0][0]
```

```python
list(myCounter.elements())
```

```python
#namedTuple section
Point = namedtuple('Point', 'x,y')
pt = Point(1,-4)
pt.x
```

```python
#OrderedDict section 
#Now it is less common since python 3.17 
exDict = OrderedDict()
exDict['a'] = 1
exDict['b'] = 5
exDict['f'] = 3
exDict['d'] = 4
exDict['e'] = 2
exDict
```

```python
#Defaultdict section
d = defaultdict(int)
d['a'] = 1 
d['b'] = 2
d['c']
```

```python
#deque section 
#can be used for double ended queue

de = deque()
de.append(22)
de.append(33)
de.appendleft(11)
de
```

```python
de.popleft()
```

```python
de.extend([44,55,66])
de
```

```python
#rotate all elements 1 place to the left
#to rorate to the right, use -1
de.rotate(1)
de
```

# Itertools 
Data types that can be used in for loops

```python
from itertools import product, permutations, combinations,  accumulate, groupby
import operator
```

```python
#product is the cartesian product of the inputs
a = [1,2]
b = [3]
prod = product(a,b)
list(prod)
```

```python
#permutations section
#all possible ordering of an input
c = [1,2,3]
perm = permutations(c)
list(perm)
```

```python
#combinations section 
#all combination with a specified length -> Combination does not include repetiton 
d = [1,2,3,4]
comb = combinations(d,2)
list(comb)
```

```python
#accumulate section
#accumulated sum of the numbers that are added in the order
acc = accumulate(d)
list(acc)
```

```python
acc_op = accumulate(d, func=operator.mul)
list(acc_op)
```

```python
e = [1,2,5,3,4]
acc_max = accumulate(e, func=max)
list(acc_max)
```

```python
#groupby section
def smaller_than_3 (x):
    return x < 3

groupObj = groupby(e, key=smaller_than_3)
for key, value in groupObj:
    print(key, list(value))
```

```python
#infinite iterable tools 
from itertools import count, repeat, cycle
```

```python
#count -> will count from a number to infite:
#cycle -> Will cycle over the numebr given infinitely
#repeat -> will repeat the same number infinitely
for i in repeat(1,4):
    print(i)
```
