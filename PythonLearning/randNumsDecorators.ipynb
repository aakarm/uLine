# Decorators
Writing our own, difference function and class decorators


```python
#Function decorator
# adding a decorator at the top of another function adds the capability to that function
# functions are first class objects and can be defined inside another function and can be passed or returned by another function 
import functools

def startEndDecorator(func):

    @functools.wraps(func)    
    def wrapper(*args, **kwargs):
        #do something before funciton func 
        print('start')
        result = func(*args, **kwargs)
        #do something after the function func
        print('end') 
        return result
    return wrapper

def printName():
    print('Aakar')

printName = startEndDecorator(printName)
printName()
```

```python
# the same extended functionality can also be acheived by

@startEndDecorator
def printAge():
    print(30)

printAge()
```

```python
#fucntion with arguments
@startEndDecorator
def add5(x):
    return x+5

#now the wrapper function in the decorator will require additional arguments
result = add5(4)
print(result)
```

```python
#stack decorators
def debug(func):
    functools.wraps(func)
    def wrapper(*args, **kwargs):
        argsRepr = [repr(a) for a in args]
        kwargsRepr = [f"{k}={v!r}" for k,v in kwargs.items()]
        signature = ", ".join(argsRepr + kwargsRepr)
        print(f"Calling {func.__name__}({signature})")
        result = func(*args, **kwargs)
        print(f"{func.__name__!r} returned {result!r}")
        return result
    return wrapper

#executed in the order they are listed
@debug
@startEndDecorator
def sayHello(name):
    greeting = f'Hello {name}'
    print(greeting)
    return greeting

sayHello('Aakar')
```

### Class decorators

```python
class CountCalls:
    def __init__(self, func):
        self.func = func
        self.num_calls = 0

    def __call__(self, *args, **kwargs):
        self.num_calls += 1
        print(f'This is executed {self.num_calls} times')
        return self.func(*args, **kwargs)

@CountCalls
def helloFunc():
    print('hello')

helloFunc()
helloFunc()
```
